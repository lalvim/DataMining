## Teste de cálculo, tabela e visualização

Usaremos uma amostra sintética com semente fixa. Para $n$ observações $x_1,\ldots,x_n$, a média amostral é

$$
\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i.
$$

Aqui, $n$ é o tamanho da amostra e $x_i$ é a observação na posição $i$.


````python
import matplotlib.pyplot as plt

RANDOM_STATE = 42
rng = np.random.default_rng(RANDOM_STATE)
amostra = rng.normal(loc=10, scale=2, size=200)

resumo = pd.Series(amostra).describe()
resumo
````

````python
sns.set_theme(style="whitegrid", context="notebook")
fig, ax = plt.subplots(figsize=(9, 5), dpi=120)
sns.histplot(amostra, bins=20, kde=True, color="#0072B2", ax=ax)
ax.axvline(amostra.mean(), color="#D55E00", linestyle="--", label="Média")
ax.set(title="Distribuição da amostra de teste", xlabel="Valor", ylabel="Frequência")
ax.legend()
plt.show()
````

A amostra foi gerada ao redor de 10. A linha tracejada permite comparar o centro calculado com a forma da distribuição. Como a geração é aleatória, a média observada não precisa ser exatamente 10; a semente garante que todos obtenham a mesma amostra.
